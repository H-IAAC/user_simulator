# **Plot IMU Data**

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import Dropdown, interact, VBox, Label
import IPython.display as display
import os

# Carrega os dados
acc_data = np.load('generated/acc.npz')
gyro_data = np.load('generated/gyro.npz')

# Lista de sensores disponíveis
sensor_names = sorted(acc_data.keys())

print(f"\033[1mSensores disponíveis ({len(sensor_names)}): {sensor_names}\033[0m")
print(f"\033[1mForma dos dados: acc/gyro = ({acc_data[sensor_names[0]].shape[0]}, 3)\033[0m")
```

```python
def plot_sensor(sensor_name):
    """Plota acelerômetro (m/s²) e giroscópio (rad/s) para o sensor selecionado."""
    global _last_fig
    
    acc = acc_data[sensor_name]
    gyro = gyro_data[sensor_name]
    
    n = acc.shape[0]
    t = np.arange(n)  # amostras (tempo em índices)
    
    acc_labels = ['Ax (m/s²)', 'Ay (m/s²)', 'Az (m/s²)']
    gyro_labels = ['Gx (rad/s)', 'Gy (rad/s)', 'Gz (rad/s)']
    
    fig, axes = plt.subplots(3, 2, figsize=(10, 10), sharex='col')
    fig.suptitle(f'Sensor: {sensor_name}', fontsize=14, fontweight='bold')
    
    # Cores
    colors = ['#e74c3c', '#2ecc71', '#3498db']
    
    # Acelerômetro — coluna 0
    for i in range(3):
        axes[i, 0].plot(t, acc[:, i], color=colors[i], linewidth=1.3)
        axes[i, 0].set_ylabel(acc_labels[i], fontsize=11, fontweight='bold')
        axes[i, 0].grid(True, alpha=0.3)
        axes[i, 0].axhline(0, color='gray', linewidth=0.5, linestyle='--')
        axes[i, 0].tick_params(axis='both', which='major', labelsize=11)
        for label in axes[i, 0].get_yticklabels() + axes[i, 0].get_xticklabels():
            label.set_fontweight('bold')
    axes[0, 0].set_title('Acelerômetro', fontsize=12, fontweight='bold')
    
    # Giroscópio — coluna 1
    for i in range(3):
        axes[i, 1].plot(t, gyro[:, i], color=colors[i], linewidth=1.3)
        axes[i, 1].set_ylabel(gyro_labels[i], fontsize=11, fontweight='bold')
        axes[i, 1].grid(True, alpha=0.3)
        axes[i, 1].axhline(0, color='gray', linewidth=0.5, linestyle='--')
        axes[i, 1].tick_params(axis='both', which='major', labelsize=11)
        for label in axes[i, 1].get_yticklabels() + axes[i, 1].get_xticklabels():
            label.set_fontweight('bold')
    axes[0, 1].set_title('Giroscópio', fontsize=12, fontweight='bold')
    
    axes[-1, 0].set_xlabel('Amostra', fontsize=11, fontweight='bold')
    axes[-1, 1].set_xlabel('Amostra', fontsize=11, fontweight='bold')
    
    plt.tight_layout()
    
    # Guarda a figura atual para exportação posterior
    _last_fig = fig
    
    plt.show()

# Dropdown interativo
dropdown = Dropdown(
    options=sensor_names,
    value='Left_hip',
    description='Sensor:',
    style={'description_width': 'initial'},
    layout={'width': '300px'}
)

interact(plot_sensor, sensor_name=dropdown);
```

```python
def export_last_plot(output_dir='plots'):
    """Exporta o último gráfico exibido para PDF."""
    
    if '_last_fig' not in globals() or _last_fig is None:
        print('Nenhum gráfico foi plotado ainda. Selecione um sensor primeiro.')
        return
    
    os.makedirs(output_dir, exist_ok=True)
    
    title = _last_fig._suptitle.get_text()
    sensor_name = title.replace('Sensor: ', '')
    
    filepath = os.path.join(output_dir, f'{sensor_name}_imu.pdf')
    _last_fig.savefig(filepath, format='pdf', bbox_inches='tight', dpi=150)
    
    print(f'PDF exportado: {os.path.abspath(filepath)}')
    
export_last_plot()
```
